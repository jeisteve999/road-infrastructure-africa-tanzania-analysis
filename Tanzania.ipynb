```python
import pandas as pd

tanzania_data = pd.read_excel('/content/Tanzania.xlsx')

tanzania_data
```

```python
tanzania_data["Rd_length"] = tanzania_data["Rd_length"].astype(str).str.replace(",", ".").astype(float)
```

```python
tanzania_data["Rd_length_km"] = tanzania_data["Rd_length"] /1000
```

```python
total_km = tanzania_data["Rd_length_km"].sum()

paved_km = tanzania_data[tanzania_data["Surface"]== "paved"]["Rd_length_km"].sum()
unpaved_km = tanzania_data[tanzania_data["Surface"]== "unpaved"]["Rd_length_km"].sum()

pct_paved = (paved_km/ total_km)*100
pct_unpaved = (unpaved_km/ total_km)*100

print(f"Total km: {total_km: .2f}")
print(f"Paved km: {paved_km: .2f} ({pct_paved:.2f}%)")
print(f"unpaved km: {unpaved_km: .2f} ({pct_unpaved:.2f}%)")
```

```python
import matplotlib.pyplot as plt

labels = ["Paved", "Unpaved"]
values = [paved_km, unpaved_km]

plt.figure()

plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Road Surface Distribution - Tanzania")

plt.axis('equal')  # Hace el círculo perfecto
plt.show()
```
